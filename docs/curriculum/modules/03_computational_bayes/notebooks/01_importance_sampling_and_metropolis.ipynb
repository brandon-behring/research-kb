# Importance sampling and Metropolis

Estimate a target expectation and then draw a Metropolis chain.


```python
import math
import random

def target(x):
    return math.exp(-0.5 * x * x)

def proposal_sample():
    return random.gauss(0, 2)

def proposal_density(x):
    sigma = 2
    return math.exp(-(x * x) / (2 * sigma * sigma))

weighted = []
for _ in range(5000):
    x = proposal_sample()
    w = target(x) / proposal_density(x)
    weighted.append((x, w))
num = sum(x * x * w for x, w in weighted)
den = sum(w for _, w in weighted)
print('Importance-sampling estimate of E[X^2]:', round(num / den, 3))

chain = []
current = 0.0
for _ in range(2000):
    proposal = current + random.gauss(0, 1)
    ratio = min(1.0, target(proposal) / target(current))
    if random.random() < ratio:
        current = proposal
    chain.append(current)
print('Metropolis chain mean:', round(sum(chain) / len(chain), 3))
```
