# Partial pooling for normal means

Demonstrate shrinkage with a simple empirical-Bayes style calculation.


```python
groups = {'A': (3, 10.0), 'B': (20, 11.0), 'C': (5, 15.0)}
global_mean = sum(n * mean for n, mean in groups.values()) / sum(n for n, _ in groups.values())
prior_strength = 8
for name, (n, mean) in groups.items():
    pooled = global_mean
    unpooled = mean
    partial = (n * mean + prior_strength * global_mean) / (n + prior_strength)
    print(name, 'pooled=', round(pooled, 2), 'unpooled=', round(unpooled, 2), 'partial=', round(partial, 2))
```
