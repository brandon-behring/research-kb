# Varying intercepts case study

Compare pooled and partially pooled group intercepts on a toy data set.


```python
group_counts = {'red': [1, 0, 1], 'blue': [0, 0, 1, 0, 0, 1, 0, 1], 'green': [1, 1]}
global_rate = sum(sum(v) for v in group_counts.values()) / sum(len(v) for v in group_counts.values())
prior_strength = 4
for name, values in group_counts.items():
    n = len(values)
    group_rate = sum(values) / n
    partial = (sum(values) + prior_strength * global_rate) / (n + prior_strength)
    print(name, 'group_rate=', round(group_rate, 3), 'partial_pooling_rate=', round(partial, 3))
```
