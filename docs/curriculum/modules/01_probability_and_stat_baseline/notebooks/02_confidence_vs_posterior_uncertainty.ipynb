# Confidence vs posterior uncertainty

Compare a Wald interval to a Beta posterior interval approximation on a grid.


```python
from math import sqrt

heads = 8
trials = 10
p_hat = heads / trials
wald = (p_hat - 1.96 * sqrt(p_hat * (1 - p_hat) / trials), p_hat + 1.96 * sqrt(p_hat * (1 - p_hat) / trials))
posterior_grid = []
for i in range(1, 1000):
    p = i / 1000
    posterior_grid.append((p, (p ** heads) * ((1 - p) ** (trials - heads))))
total = sum(weight for _, weight in posterior_grid)
posterior_grid = [(p, weight / total) for p, weight in posterior_grid]
cdf = 0
lower = upper = None
for p, weight in posterior_grid:
    cdf += weight
    if lower is None and cdf >= 0.025:
        lower = p
    if upper is None and cdf >= 0.975:
        upper = p
        break
print('Wald interval:', tuple(round(x, 3) for x in wald))
print('Posterior interval (uniform prior):', (round(lower, 3), round(upper, 3)))
```
