# Bernoulli likelihoods

Compute the Bernoulli likelihood on a parameter grid.


```python
from math import comb

heads = 8
trials = 10
grid = [i / 100 for i in range(1, 100)]
likelihood = [comb(trials, heads) * (p ** heads) * ((1 - p) ** (trials - heads)) for p in grid]
mle = grid[likelihood.index(max(likelihood))]
print('MLE for p:', round(mle, 2))
for p, ll in list(zip(grid, likelihood))[::10]:
    print(round(p, 2), round(ll, 6))
```
