# Posterior predictive coin flips

Estimate a posterior predictive probability under a Beta-Binomial model.


```python
alpha_post, beta_post = 10, 4
p_next_head = alpha_post / (alpha_post + beta_post)
print('Posterior predictive P(next head) =', round(p_next_head, 3))
for future_heads in range(6):
    print('Future heads in 5 flips =', future_heads, '| rough probability mass placeholder for extension')
```
