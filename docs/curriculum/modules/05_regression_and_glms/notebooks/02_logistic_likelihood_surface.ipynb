# Logistic likelihood surface

Compute a coarse logistic-regression likelihood surface.


```python
import math

x = [-2, -1, 0, 1, 2]
y = [0, 0, 0, 1, 1]
for intercept in [-1.0, 0.0, 1.0]:
    for slope in [0.5, 1.0, 1.5]:
        score = 0
        for xi, yi in zip(x, y):
            eta = intercept + slope * xi
            p = 1 / (1 + math.exp(-eta))
            score += yi * math.log(p) + (1 - yi) * math.log(1 - p)
        print('intercept=', intercept, 'slope=', slope, 'loglik=', round(score, 3))
```
