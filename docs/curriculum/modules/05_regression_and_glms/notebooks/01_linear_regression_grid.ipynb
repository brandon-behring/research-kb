# Linear regression grid

Evaluate a simple linear regression likelihood over a slope grid.


```python
x = [0, 1, 2, 3, 4]
y = [1.1, 2.0, 2.9, 4.1, 5.2]
slopes = [i / 10 for i in range(0, 30)]
intercept = 1.0
sigma2 = 0.05
ll = []
for b in slopes:
    score = 0
    for xi, yi in zip(x, y):
        mu = intercept + b * xi
        score += -((yi - mu) ** 2) / (2 * sigma2)
    ll.append(score)
best = slopes[ll.index(max(ll))]
print('Best slope on grid:', best)
```
