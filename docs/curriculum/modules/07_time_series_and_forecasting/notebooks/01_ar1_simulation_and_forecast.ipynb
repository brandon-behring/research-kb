# AR(1) simulation and forecast

Simulate an AR(1) series and compute a one-step forecast.


```python
import random

phi = 0.7
series = [0.0]
for _ in range(1, 20):
    series.append(phi * series[-1] + random.gauss(0, 1))
forecast = phi * series[-1]
print('Last 5 observations:', [round(x, 3) for x in series[-5:]])
print('One-step forecast:', round(forecast, 3))
```
