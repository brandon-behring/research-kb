# Local-level state-space updates

Run a simple manual Kalman-style update for a local-level model.


```python
observations = [10.0, 10.5, 9.8, 10.2]
level_mean = 10.0
level_var = 1.0
obs_var = 0.25
state_var = 0.1
for y in observations:
    pred_mean = level_mean
    pred_var = level_var + state_var
    kalman_gain = pred_var / (pred_var + obs_var)
    level_mean = pred_mean + kalman_gain * (y - pred_mean)
    level_var = (1 - kalman_gain) * pred_var
    print('updated mean=', round(level_mean, 3), 'updated var=', round(level_var, 3))
```
