# Inventory audit

Load the curriculum registry and summarize source states, roles, and module assignments.


```python
from pathlib import Path
import json
from collections import Counter

registry_path = Path('docs/curriculum/material_registry.json')
registry = json.loads(registry_path.read_text())
materials = registry['materials']
state_counts = Counter(item['source_state'] for item in materials)
role_counts = Counter(item['role'] for item in materials)
print('Source states:', dict(state_counts))
print('Roles:', dict(role_counts))
print('Core materials:')
for item in materials:
    if item['role'] == 'core':
        print('-', item['title'], '|', item['source_state'])
```
