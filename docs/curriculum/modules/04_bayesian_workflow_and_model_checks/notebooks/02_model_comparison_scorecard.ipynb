# Model comparison scorecard

Use held-out log score as a lightweight stand-in before full LOO/WAIC tooling.


```python
import math

train_heads, train_trials = 8, 10
test_heads, test_trials = 3, 5
models = {'fixed_0.5': 0.5, 'posterior_mean': 10 / 14}
for name, p in models.items():
    log_score = test_heads * math.log(p) + (test_trials - test_heads) * math.log(1 - p)
    print(name, 'held-out log score =', round(log_score, 3))
print('Use ArviZ/PSIS-LOO for the real workflow once probabilistic-programming models are in place.')
```
