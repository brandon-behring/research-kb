# Poisson posterior predictive check

Fit a Gamma-Poisson update and simulate posterior predictive draws.


```python
import random

counts = [3, 4, 2, 5, 4, 3, 6, 2]
alpha_post = 1 + sum(counts)
beta_post = 1 + len(counts)
posterior_mean = alpha_post / beta_post
print('Posterior mean lambda:', round(posterior_mean, 3))
pp_draws = []
for _ in range(200):
    lam = random.gammavariate(alpha_post, 1 / beta_post)
    running = 0
    t = 0
    while running < lam:
        t += 1
        running += random.expovariate(1.0)
    pp_draws.append(max(t - 1, 0))
print('Observed max count:', max(counts))
print('Posterior predictive average max proxy:', round(sum(pp_draws) / len(pp_draws), 3))
```
