# Variational objective demo

Compare moments of a target Gaussian and an approximating Gaussian as a simple ELBO intuition builder.


```python
target = {'mean': 0.0, 'var': 1.0}
approx = {'mean': 0.5, 'var': 1.5}
mean_gap = approx['mean'] - target['mean']
var_gap = approx['var'] - target['var']
print('Target moments:', target)
print('Approximate moments:', approx)
print('Mean gap:', mean_gap)
print('Variance gap:', var_gap)
print('Next step: connect this mismatch to KL direction and ELBO optimization in ProbML / AEVB.')
```
