# Graphical model factorization

Factor a simple DAG into conditional distributions.


```python
dag = {'A': [], 'B': ['A'], 'C': ['A'], 'D': ['B', 'C']}
for node, parents in dag.items():
    if parents:
        print(f'p({node} | {', '.join(parents)})')
    else:
        print(f'p({node})')
print('Joint factorization: p(A) p(B|A) p(C|A) p(D|B,C)')
```
